Convert to Lowercase

```python
text = "Tokenization is the process of breaking text into smaller units called tokens, which helps machines process and analyze language effectively."
text = text.lower()

print(text)
```

Remove Punctuation

```python
import string

text = "Tokenization is the process of breaking text into smaller units called tokens, which helps machines process and analyze language effectively."
text = text.translate(str.maketrans("", "", string.punctuation))

print(text)
```

Tokenization

```python
from nltk.tokenize import word_tokenize

text = "Tokenization is the process of breaking text into smaller units called tokens, which helps machines process and analyze language effectively."
tokens = word_tokenize(text)

print(tokens)
```

Stop Word Removal

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

text = "Tokenization is the process of breaking text into smaller units called tokens, which helps machines process and analyze language effectively."
words = text.lower().split()

result = [word for word in words if word not in stop_words]

print(result)
```

Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays"]
print([stemmer.stem(word) for word in words])
```

Lemmatization

```python
import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp("Tokenization is the process of breaking text into smaller units called tokens, which helps machines process and analyze language effectively.")

for token in doc:
    print(token.text, "->", token.lemma_)
```

Remove Numbers

```python
import re

text = "I have 2 apples and 10 oranges."

text = re.sub(r'\d+', '', text)

print(text)
```

Remove Extra Spaces

```python
text = "I    love     NLP"

text = " ".join(text.split())

print(text)
```
